```python

from google.colab import files
import pandas as pd
import numpy as np
import io

uploaded = files.upload()

train = pd.read_csv(io.BytesIO(uploaded['train.csv']))
test = pd.read_csv(io.BytesIO(uploaded['test.csv']))

print("Training data shape:", train.shape)
print("Test data shape:", test.shape)

print("\nTraining columns:")
print(train.columns.tolist())

print("\nTest columns:")
print(test.columns.tolist())
```

```python

# 1. Data types and missing values
print("=== TRAIN DATA INFO ===")
train.info()

print("\n=== MISSING VALUES ===")
print("Train:\n", train.isnull().sum())
print("\nTest:\n", test.isnull().sum())

# 2. Duplicate records
print("\n=== DUPLICATES ===")
print("Duplicate training rows:", train.duplicated().sum())
print("Duplicate IDs in train:", train['id'].duplicated().sum())
print("Duplicate IDs in test:", test['id'].duplicated().sum())

# 3. Target variable distribution
print("\n=== PRICE STATISTICS ===")
print(train['price'].describe())

# 4. Number of unique values in each feature
print("\n=== UNIQUE VALUES ===")
print(train.nunique().sort_values())

# 5. Sample records
print("\n=== FIRST 5 TRAINING ROWS ===")
display(train.head())

print("\n=== FIRST 5 TEST ROWS ===")
display(test.head())
```

```python

# Check zero and unusually high prices
print("Zero-price rows:", (train['price'] == 0).sum())
print("Prices above 1000:", (train['price'] > 1000).sum())
print("Prices above 2000:", (train['price'] > 2000).sum())

print("\n=== HIGHEST-PRICE RECORDS ===")
display(train.nlargest(10, 'price')[
    ['airline', 'flight', 'source_city', 'destination_city',
     'flight_class', 'stops', 'duration', 'days_left', 'price']
])

print("\n=== ZERO-PRICE RECORDS ===")
display(train.loc[train['price'] == 0, [
    'airline', 'flight', 'source_city', 'destination_city',
    'flight_class', 'stops', 'duration', 'days_left', 'price'
]].head(10))

# Categorical values and their frequencies
categorical_cols = train.select_dtypes(include='object').columns.tolist()

for col in categorical_cols:
    print(f"\n=== {col} ({train[col].nunique()} unique values) ===")
    print(train[col].value_counts().head(15))
```

```python

# Install XGBoost only if it is not already installed
import importlib.util
import subprocess
import sys

if importlib.util.find_spec("xgboost") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "xgboost"
    ])

import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Separate features and target
X = train.drop(columns=["price", "id"])
y = train["price"]

# Identify feature types
categorical_cols = X.select_dtypes(include=["object", "category"]).columns.tolist()
numerical_cols = X.select_dtypes(include=["number"]).columns.tolist()

print("Categorical features:", categorical_cols)
print("Numerical features:", numerical_cols)

# Split the data for validation
X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
        ("num", "passthrough", numerical_cols)
    ]
)

# Baseline XGBoost model
xgb_model = xgb.XGBRegressor(
    objective="reg:squarederror",
    n_estimators=500,
    learning_rate=0.05,
    max_depth=8,
    min_child_weight=3,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=5,
    random_state=42,
    n_jobs=-1,
    tree_method="hist"
)

baseline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", xgb_model)
])

# Train
baseline.fit(X_train, y_train)

# Validate
val_pred = baseline.predict(X_val)

mae = mean_absolute_error(y_val, val_pred)
rmse = np.sqrt(mean_squared_error(y_val, val_pred))
r2 = r2_score(y_val, val_pred)

print("\n=== BASELINE XGBOOST RESULTS ===")
print(f"MAE:  {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²:   {r2:.4f}")
```

```python

import joblib
from sklearn.model_selection import RandomizedSearchCV

# Save the baseline pipeline, including preprocessing
joblib.dump(baseline, "xgboost_baseline.joblib")
print("Baseline model saved: xgboost_baseline.joblib")

# Hyperparameter search space
param_distributions = {
    "model__n_estimators": [300, 500, 700, 900],
    "model__learning_rate": [0.02, 0.03, 0.05, 0.07, 0.1],
    "model__max_depth": [3, 4, 5, 6, 8],
    "model__min_child_weight": [1, 3, 5, 7],
    "model__subsample": [0.7, 0.8, 0.9, 1.0],
    "model__colsample_bytree": [0.7, 0.8, 0.9, 1.0],
    "model__reg_lambda": [1, 3, 5, 10, 15],
    "model__reg_alpha": [0, 0.01, 0.1, 1, 5],
    "model__gamma": [0, 0.1, 0.3, 0.5, 1]
}

# Randomized hyperparameter tuning
random_search = RandomizedSearchCV(
    estimator=baseline,
    param_distributions=param_distributions,
    n_iter=15,
    scoring="neg_mean_absolute_error",
    cv=3,
    verbose=1,
    random_state=42,
    n_jobs=1,
    return_train_score=True
)

random_search.fit(X_train, y_train)

# Best model
best_model = random_search.best_estimator_

print("\n=== BEST HYPERPARAMETERS ===")
print(random_search.best_params_)

print("\nBest 3-fold CV MAE:", round(-random_search.best_score_, 4))

# Evaluate on the same held-out validation set
tuned_val_pred = best_model.predict(X_val)

tuned_mae = mean_absolute_error(y_val, tuned_val_pred)
tuned_rmse = np.sqrt(mean_squared_error(y_val, tuned_val_pred))
tuned_r2 = r2_score(y_val, tuned_val_pred)

print("\n=== TUNED XGBOOST VALIDATION RESULTS ===")
print(f"MAE:  {tuned_mae:.4f}")
print(f"RMSE: {tuned_rmse:.4f}")
print(f"R²:   {tuned_r2:.4f}")

# Compare with baseline
print("\n=== BASELINE VS TUNED ===")
print(f"Baseline MAE: {mae:.4f} | Tuned MAE: {tuned_mae:.4f}")
print(f"Baseline RMSE: {rmse:.4f} | Tuned RMSE: {tuned_rmse:.4f}")
print(f"Baseline R²: {r2:.4f} | Tuned R²: {tuned_r2:.4f}")

# Save the tuned pipeline
joblib.dump(best_model, "xgboost_tuned.joblib")
print("\nTuned model saved: xgboost_tuned.joblib")
```

```python

# Retrain the selected model on the complete training dataset
final_model = best_model
final_model.fit(X, y)

# Save the final trained pipeline, including preprocessing
joblib.dump(final_model, "xgboost_final_model.joblib")

print("Final model saved: xgboost_final_model.joblib")

# Prepare test features in the same format as training
X_test = test.drop(columns=["id"])

# Predict flight prices
test_predictions = final_model.predict(X_test)

# Prices cannot be negative
test_predictions = np.maximum(test_predictions, 0)

# Create Kaggle submission in the original test-row order
submission = pd.DataFrame({
    "id": test["id"].values,
    "price": test_predictions
})

# Validate the submission format
assert submission.shape == (4520, 2), f"Unexpected shape: {submission.shape}"
assert submission.columns.tolist() == ["id", "price"]
assert submission["id"].equals(test["id"])
assert submission["price"].notna().all()

# Save submission
submission.to_csv("submission.csv", index=False)

print("\nSubmission saved: submission.csv")
print("Submission shape:", submission.shape)
print("\nFirst 10 predictions:")
display(submission.head(10))

print("\nFile and format checks passed!")
```

```python

import importlib.util
import subprocess
import sys

if importlib.util.find_spec("catboost") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "catboost"
    ])

from catboost import CatBoostRegressor

# Prepare categorical features as strings
X_train_cb = X_train.copy()
X_val_cb = X_val.copy()

for col in categorical_cols:
    X_train_cb[col] = X_train_cb[col].astype(str)
    X_val_cb[col] = X_val_cb[col].astype(str)

# Train CatBoost
catboost_model = CatBoostRegressor(
    iterations=1500,
    learning_rate=0.05,
    depth=8,
    loss_function="RMSE",
    eval_metric="MAE",
    l2_leaf_reg=5,
    random_seed=42,
    verbose=100,
    early_stopping_rounds=100,
    allow_writing_files=False
)

catboost_model.fit(
    X_train_cb,
    y_train,
    cat_features=categorical_cols,
    eval_set=(X_val_cb, y_val),
    verbose=100
)

# Validation predictions
catboost_pred = catboost_model.predict(X_val_cb)

catboost_mae = mean_absolute_error(y_val, catboost_pred)
catboost_rmse = np.sqrt(mean_squared_error(y_val, catboost_pred))
catboost_r2 = r2_score(y_val, catboost_pred)

print("\n=== CATBOOST VALIDATION RESULTS ===")
print(f"MAE:  {catboost_mae:.4f}")
print(f"RMSE: {catboost_rmse:.4f}")
print(f"R²:   {catboost_r2:.4f}")

# Save the trained CatBoost model
catboost_model.save_model("catboost_model.cbm")
print("\nCatBoost model saved: catboost_model.cbm")
```

```python

from catboost import CatBoostRegressor
import joblib

# Candidate hyperparameter configurations
configs = [
    {"depth": 6, "learning_rate": 0.03, "l2_leaf_reg": 5},
    {"depth": 7, "learning_rate": 0.03, "l2_leaf_reg": 5},
    {"depth": 8, "learning_rate": 0.03, "l2_leaf_reg": 7},
    {"depth": 8, "learning_rate": 0.05, "l2_leaf_reg": 10},
]

catboost_results = []
best_catboost = None
best_catboost_mae = float("inf")

for i, params in enumerate(configs, start=1):
    print(f"\n{'='*15} Configuration {i}/{len(configs)} {'='*15}")
    print(params)

    model = CatBoostRegressor(
        iterations=2500,
        loss_function="RMSE",
        eval_metric="MAE",
        random_seed=42,
        verbose=False,
        early_stopping_rounds=150,
        allow_writing_files=False,
        **params
    )

    model.fit(
        X_train_cb,
        y_train,
        cat_features=categorical_cols,
        eval_set=(X_val_cb, y_val),
        verbose=250
    )

    predictions = model.predict(X_val_cb)

    current_mae = mean_absolute_error(y_val, predictions)
    current_rmse = np.sqrt(mean_squared_error(y_val, predictions))
    current_r2 = r2_score(y_val, predictions)

    print("Best iteration:", model.get_best_iteration())
    print(f"Validation MAE:  {current_mae:.4f}")
    print(f"Validation RMSE: {current_rmse:.4f}")
    print(f"Validation R²:   {current_r2:.4f}")

    catboost_results.append({
        "configuration": i,
        "depth": params["depth"],
        "learning_rate": params["learning_rate"],
        "l2_leaf_reg": params["l2_leaf_reg"],
        "best_iteration": model.get_best_iteration(),
        "MAE": current_mae,
        "RMSE": current_rmse,
        "R2": current_r2
    })

    if current_mae < best_catboost_mae:
        best_catboost_mae = current_mae
        best_catboost = model

# Display ranking
results_df = pd.DataFrame(catboost_results).sort_values("MAE")
print("\n=== CATBOOST TUNING RESULTS (RANKED BY MAE) ===")
display(results_df)

# Save the best tuned CatBoost model
best_catboost.save_model("catboost_tuned.cbm")
print("Best tuned CatBoost model saved: catboost_tuned.cbm")

print("\nPrevious CatBoost MAE:", round(catboost_mae, 4))
print("Best tuned CatBoost MAE:", round(best_catboost_mae, 4))
```

```python

# Prepare test data for CatBoost
X_test_cb = test.drop(columns=["id"]).copy()

for col in categorical_cols:
    X_test_cb[col] = X_test_cb[col].astype(str)

# Predict test prices using the original CatBoost model
catboost_test_predictions = catboost_model.predict(X_test_cb)

# Ensure predictions are non-negative
catboost_test_predictions = np.maximum(
    catboost_test_predictions, 0
)

# Create submission
catboost_submission = pd.DataFrame({
    "id": test["id"].values,
    "price": catboost_test_predictions
})

# Validate and save
assert catboost_submission.shape == (4520, 2)
assert catboost_submission.columns.tolist() == ["id", "price"]
assert catboost_submission["id"].equals(test["id"])
assert catboost_submission["price"].notna().all()

catboost_submission.to_csv("catboost.csv", index=False)

print("Saved: catboost.csv")
print("Shape:", catboost_submission.shape)
display(catboost_submission.head())

# Download the CSV to your computer
from google.colab import files
files.download("catboost.csv")
```

```python

# Validation predictions from the existing models
xgb_val_pred = best_model.predict(X_val)
cat_val_pred = catboost_model.predict(X_val_cb)

# Test different ensemble weights
ensemble_results = []

for cat_weight in [0.3, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9]:
    xgb_weight = 1 - cat_weight

    ensemble_pred = (
        cat_weight * cat_val_pred
        + xgb_weight * xgb_val_pred
    )

    ensemble_mae = mean_absolute_error(y_val, ensemble_pred)
    ensemble_rmse = np.sqrt(mean_squared_error(y_val, ensemble_pred))
    ensemble_r2 = r2_score(y_val, ensemble_pred)

    ensemble_results.append({
        "catboost_weight": cat_weight,
        "xgboost_weight": xgb_weight,
        "MAE": ensemble_mae,
        "RMSE": ensemble_rmse,
        "R2": ensemble_r2
    })

ensemble_results_df = pd.DataFrame(ensemble_results).sort_values("MAE")

print("=== ENSEMBLE VALIDATION RESULTS ===")
display(ensemble_results_df)

# Select the best weights based on validation MAE
best_weights = ensemble_results_df.iloc[0]
best_cat_weight = best_weights["catboost_weight"]
best_xgb_weight = best_weights["xgboost_weight"]

best_ensemble_val_pred = (
    best_cat_weight * cat_val_pred
    + best_xgb_weight * xgb_val_pred
)

print("\nBest CatBoost weight:", best_cat_weight)
print("Best XGBoost weight:", best_xgb_weight)
print(f"Best Ensemble MAE:  {mean_absolute_error(y_val, best_ensemble_val_pred):.4f}")
print(f"Best Ensemble RMSE: {np.sqrt(mean_squared_error(y_val, best_ensemble_val_pred)):.4f}")
print(f"Best Ensemble R²:   {r2_score(y_val, best_ensemble_val_pred):.4f}")

print(f"\nCatBoost-only MAE: {mean_absolute_error(y_val, cat_val_pred):.4f}")
print(f"XGBoost-only MAE:  {mean_absolute_error(y_val, xgb_val_pred):.4f}")
```

```python

# Recalculate predictions from the exact models used in the ensemble

xgb_check_pred = best_model.predict(X_val)
cat_check_pred = catboost_model.predict(X_val_cb)

ensemble_check_pred = (
    0.3 * cat_check_pred
    + 0.7 * xgb_check_pred
)

# Print metrics side by side
comparison = []

for name, pred in [
    ("XGBoost", xgb_check_pred),
    ("CatBoost", cat_check_pred),
    ("Ensemble (30% CatBoost, 70% XGBoost)", ensemble_check_pred)
]:
    comparison.append({
        "Model": name,
        "MAE": mean_absolute_error(y_val, pred),
        "RMSE": np.sqrt(mean_squared_error(y_val, pred)),
        "R2": r2_score(y_val, pred)
    })

comparison_df = pd.DataFrame(comparison).sort_values("MAE")
display(comparison_df.round(4))
```
